```python

# Importações
import pandas as pd
import re
from google import genai
from google.colab import userdata, files
```

```python
# Lê sua chave salva nos Secrets do Colab
GEMINI_API_KEY = userdata.get("GeminiAPIKeyAulaPy04")

# Conecta ao Gemini
client = genai.Client(api_key=GEMINI_API_KEY)
```

```python



# Lê o arquivo com 13 avaliações
df_reviews = pd.read_csv("/Dados/reviews_13_exemplos.csv")

# Junta todas as avaliações para enviar em UMA requisição
avaliacoes = "\n\n".join(
    f"{i + 1}. {texto}"
    for i, texto in enumerate(df_reviews["reviewText"])
)

# Instrução para a IA
prompt = f"""
Classifique cada avaliação como Positiva, Negativa ou Neutra.

Responda exatamente uma classificação por linha, na mesma ordem
das avaliações. Não explique, não numere e não escreva mais nada.

Avaliações:
{avaliacoes}
"""

# Faz somente uma chamada para a IA
resposta = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

# Extrai somente Positiva, Negativa ou Neutra da resposta
classificacoes = re.findall(
    r"\b(positiva|negativa|neutra)\b",
    resposta.text,
    flags=re.IGNORECASE
)

classificacoes = [item.capitalize() for item in classificacoes]

# Confere se a IA classificou todas as 13 avaliações
if len(classificacoes) != len(df_reviews):
    print(resposta.text)
    raise ValueError(
        f"Foram encontradas {len(classificacoes)} classificações, "
        f"mas existem {len(df_reviews)} avaliações."
    )

# Cria a nova coluna e salva o arquivo final
df_reviews["classificacao_ia"] = classificacoes

caminho_saida = "/Dados/reviews_13_com_respostas.csv"

df_reviews.to_csv(
    caminho_saida,
    index=False,
    encoding="utf-8-sig"
)

# Mostra o resultado
display(df_reviews[["reviewText", "classificacao_ia"]])

# Baixa o arquivo para seu computador
files.download(caminho_saida)

print("Concluído!")
```
