```python
import os
from google.colab import userdata
os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY_2')
```

```python
from google import genai
client = genai.Client()
```

```python
resposta = client.interactions.create(model='gemini-3.5-flash', input='O que é IA?')
print(resposta.output_text)
```
