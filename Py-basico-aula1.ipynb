```python
print("### 1. Olá Mundo e Variáveis Básicas ###")
# Utilizando o comando print para exibir números e textos.
print("Olá, Mundo!")
print(12345)

# Diferenciando números e strings aplicando aspas em textos.
print("Número como string: " + "100") # String
print(100) # Número

# Criando variáveis para armazenar valores e texto.
my_string = "Este é um texto."
my_number = 25
print("Variável de texto:", my_string)
print("Variável numérica:", my_number)

print("\n### 2. Concatenação e Conversão de Tipos ###")
# Concatene strings e converta tipos quando necessário.
string_parte1 = "Python é "
string_parte2 = "incrível!"
concatenated_string = string_parte1 + string_parte2
print(concatenated_string)

# Tentando concatenar string e número (erro sem conversão)
# print("Meu número é: " + my_number) # Isso causaria um TypeError
print("Meu número é: " + str(my_number)) # Conversão explícita para string

print("\n### 3. Operações Aritméticas e Variáveis ###")
# Realize operações de soma, subtração, multiplicação e divisão.
num1 = 50
num2 = 10

# Soma
soma = num1 + num2
print(f"{num1} + {num2} = {soma}")

# Subtração
subtracao = num1 - num2
print(f"{num1} - {num2} = {subtracao}")

# Multiplicação
multiplicacao = num1 * num2
print(f"{num1} * {num2} = {multiplicacao}")

# Divisão
divisao = num1 / num2
print(f"{num1} / {num2} = {divisao}")

# Utilizando variáveis para calcular totais e dividir custos.
custo_total = 150.75
numero_pessoas = 3
divisao_custo = custo_total / numero_pessoas
print(f"Custo total: R${custo_total:.2f}, dividido por {numero_pessoas} pessoas: R${divisao_custo:.2f} por pessoa.")

# Guardando resultados de operações em novas variáveis.
resultado_final = (soma + subtracao) * divisao
print("Resultado final de operações aninhadas:", resultado_final)

print("\n### 4. Métodos de String ###")
# Aplique métodos lower, upper, strip e replace em textos.
texto_original = "  Olá, MUNDO Python!  "
print(f"Texto Original: '{texto_original}'")

# lower()
texto_lower = texto_original.lower()
print(f"Após lower(): '{texto_lower}'")

# upper()
texto_upper = texto_original.upper()
print(f"Após upper(): '{texto_upper}'")

# strip()
texto_strip = texto_original.strip()
print(f"Após strip(): '{texto_strip}'")

# replace()
texto_replace = texto_original.replace("MUNDO", "Universo")
print(f"Após replace('MUNDO', 'Universo'): '{texto_replace}'")

# Atualize strings e sobrescreva variáveis conforme o resultado.
texto_original = texto_original.strip().replace("MUNDO", "Python").lower()
print(f"Texto atualizado (strip, replace, lower): '{texto_original}'")

print("\n### 5. Interação com o Usuário (Input) e f-strings ###")
# Capture dados do usuário usando input com mensagens.
# Converta entradas para tipos adequados, como int, quando preciso.

nome = input("Por favor, digite seu nome: ")
idade_str = input("Agora, digite sua idade: ")

# Utilizando f-string para concatenar variáveis e textos facilmente.
print(f"Olá, {nome}! Parece que você tem {idade_str} anos.")

# Convertendo a idade para inteiro para operações numéricas
try:
    idade = int(idade_str)
    print(f"Sua idade como número inteiro é: {idade}")
    print(f"No próximo ano, você terá {idade + 1} anos.")
except ValueError:
    print(f"'{idade_str}' não é uma idade válida. Por favor, digite um número.")
    idade = 0 # Define uma idade padrão para evitar erros futuros

print("\n### 6. Estruturas Condicionais (if, elif, else) ###")
# Implemente estruturas condicionais para verificações.
# Mantenha a identação correta em blocos condicionais.
# Ajuste condições para validar idades e definir fluxos (aprovado, recuperação, reprovado).

# Exemplo 1: Verificação simples de idade
if idade >= 18:
    print(f"Com {idade} anos, você é maior de idade.")
else:
    print(f"Com {idade} anos, você é menor de idade.")

# Exemplo 2: Sistema de notas (aprovado, recuperação, reprovado)
nota_aluno_str = input("Digite a nota do aluno (0-100): ")
try:
    nota_aluno = int(nota_aluno_str)
    if 0 <= nota_aluno <= 100: # Validação de entrada
        if nota_aluno >= 70:
            print(f"Aluno com nota {nota_aluno}: APROVADO!")
        elif nota_aluno >= 50:
            print(f"Aluno com nota {nota_aluno}: Em RECUPERAÇÃO.")
        else:
            print(f"Aluno com nota {nota_aluno}: REPROVADO.")
    else:
        print(f"Nota {nota_aluno} inválida. Por favor, digite um número entre 0 e 100.")
except ValueError:
    print(f"'{nota_aluno_str}' não é uma nota válida. Por favor, digite um número.")
```
